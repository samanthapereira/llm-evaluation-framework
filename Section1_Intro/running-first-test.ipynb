```python
from dotenv import load_dotenv
from pathlib import Path

import os
load_dotenv("../.env", override=True)
openai_key = os.getenv("OPENAI_API_KEY")
confident_key = os.getenv("CONFIDENT_API_KEY")
env_path = Path("../.env").resolve()

print("Environment file:", env_path)
print("File exists:", env_path.exists())
print("OpenAI API key configured:", bool(openai_key))
print("Confident AI API key configured:", bool(confident_key))
```

```python
from deepeval.test_case import LLMTestCase
from deepeval.metrics import AnswerRelevancyMetric
from deepeval.metrics import FaithfulnessMetric
from deepeval.metrics import ContextualPrecisionMetric

answer_relevancy_metric = AnswerRelevancyMetric()
faithfullness_metric = FaithfulnessMetric()
contextual_precision_metric = ContextualPrecisionMetric()
api_key = os.getenv("OPENAI_API_KEY")
test_case = LLMTestCase(
    input = "Who is the current President of United states",
    # Should come from an LLM or from an Agent or RAG
    actual_output = "Donald Trump",
    # RAG - Vector DB, AI Agent -Agent tools, LLM- LLM Invoke response
    retrieval_context=["Donald Trump is the current president"],
    expected_output="Trump is the current president"
)

answer_relevancy_metric.measure(test_case)
print(answer_relevancy_metric.score)
faithfullness_metric.measure(test_case)
print(faithfullness_metric.score)
contextual_precision_metric.measure(test_case=test_case)
print(contextual_precision_metric.score)
```
